## Bloco 1 — Preparação dos dados

```python
from data_pipe.data_prep import prepare, frame_ids, get_frame

# Carrega os dados de exemplo da Metrica Sports via kloppy
# limit=None carrega a partida inteira (~2000 frames por período)
# Para testes rápidos use limit=500
df = prepare(
    match_id=1,       # 1 ou 2 (Metrica open data)
    limit=500,        # None = partida completa
    smooth_window=3   # janela do filtro de suavização (frames)
)

print(f'Frames carregados : {len(frame_ids(df))}')
print(f'Linhas no DataFrame: {len(df)}')
df.head()
```

## Bloco 2 — Visualização interativa

```python
# Para o slider funcionar no Jupyter é necessário o backend interativo
# No VS Code / JupyterLab use: %matplotlib widget
# No Jupyter classic use:      %matplotlib tk
%matplotlib widget

import vis
from vis.player import interactive, snapshot

interactive(
    df,
    show_skeleton=True,   # linhas entre vizinhos mais próximos
    show_delaunay=True,   # Shape Graph (Triangulação de Delaunay)
    show_hull=True        # polígono convexo de cada time
)
```

## Extra — Exportar um frame como imagem estática

```python
%matplotlib inline
from apagr.viz import snapshot

ids = frame_ids(df)

# Salva o frame do minuto ~2 como PNG
snapshot(df, ids[50], path='frame_50.png')
```
